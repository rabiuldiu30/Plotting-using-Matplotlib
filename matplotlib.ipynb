```python
# import the  library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import  seaborn as sns

plt.style.use('ggplot')
```

```python
# plotting a simple function
price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]

plt.style.use('ggplot')

plt.plot(year,price)
plt.title("Simple Function")
plt.xlabel('Year')
plt.ylabel('Price')
plt.show()
```

```python
# from  a pandas dataframe
batsman = pd.read_csv('sharma-kohli.csv')
batsman

plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashdot',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',linewidth=2)

plt.title('Rohit Sharma vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='yellow',linestyle='dashdot',linewidth=3,marker='D',markersize='10')
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue',linestyle='dashdot',linewidth=2,marker='o',markersize='5')

plt.title('Rohit Sharma vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='yellow',linestyle='dashdot',linewidth=3,marker='D',markersize='10',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue',linestyle='dashdot',linewidth=2,marker='o',markersize='5',label='Rohit')

plt.title('Rohit Sharma vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend()
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='yellow',linestyle='dashdot',linewidth=3,marker='D',markersize='10',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue',linestyle='dashdot',linewidth=2,marker='o',markersize='5',label='Rohit')

plt.title('Rohit Sharma vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend(loc='upper right')
```
