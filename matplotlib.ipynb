```python
# import the  library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import  seaborn as sns

plt.style.use('ggplot')
```

```python
# plotting a simple function
price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]

plt.style.use('ggplot')

plt.plot(year,price)
plt.title("Simple Function")
plt.xlabel('Year')
plt.ylabel('Price')
plt.show()
```

```python
# size
tips = sns.load_dataset('tips')

# slower

plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
```

```python
# scatterplot using plt.plot
# faster
plt.plot(tips['total_bill'],tips['tip'],'o')
```

Bar chart

```python
# a simple graph
children = [10,20,30,10,25]
colors = ['red','green','blue','pink','yellow']

plt.barh(colors,children,color='skyblue')
```

```python
# color and label
df =  pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'])
```

```python
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'])
plt.bar(df['batsman'],df['2017'])
```

```python
plt.bar(np.arange(df.shape[0]) - 0.5,df['2015'])
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2,color='skyblue')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='orange')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color='red')

plt.xticks(np.arange(df.shape[0]),df['batsman'])

plt.show()
```

```python
# Stacked Bar Chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=df['2016'],label='2015')

plt.legend()
plt.show()
```
